```python
#to repeat any array for number of times

x=np.array([[1,2,3]])   #give it in 2d so it can run in axis=1 too( [ [] ] )
arr1=np.repeat(x,3)
arr2=np.repeat(x,3, axis=0)   #axis=0: The Vertical direction (Downwards, along the Rows)
arr3=np.repeat(x,4, axis=1)   #axis=1: The Horizontal direction (Sideways, along the Columns)
print(arr1)
print(arr2)
print(arr3)
```

```python
######## COPYING OF ARRAYS--- IMPORTANT

m=np.array([1,2,3,4])
n=m
print('m=',m)
print('n=',n)
n[0]=100
print('m=',m)   #change in n AFFECTED m too
print('n=',n)

```

```python
###### TO AVOID THAT MISTAKE FROM PREVIOUS COPYING CODE:

p=np.array([1,2,3,4])
q=p.copy()
print('p=',p)
print('q=',q)
q[0]=100
print('p=',p)     # change in q DID NOT AFFECT p
print('q=',q)
```

```python
# Arithmetic and basic math operations

print(m+2)
print(m-2)
print(m*2)
print(m/2)
print(m**2)
print(m+p)
print(m-p)
print(np.cos(m))
```

```python
###Linear algebra           ### [  https://docs.scipy.org/doc/numpy/reference/routines.linalg.html  ]

a=np.ones((2,3))
print(a)
b=np.full((3,2),2)
print(b)
print(np.matmul(a,b))       #to multiply matrix and simply using a*b gives error
c=np.identity(3)
print(np.linalg.det(c))     #REMEMBER, determinant can be found only for square matrices
```

```python
###Statistics

stat=np.array([[1,2,3],[4,5,6]])
print(np.max(stat))          #to get max number of matrix
print(np.max(stat,axis=0))   #to get max number along the rows,ie,downward
print(np.max(stat,axis=1))   #to get max number along the columns,ie,sideways
print(np.min(stat))          #to get min number of matrix
print(np.min(stat,axis=0))   #to get min number along the rows,ie,downward
print(np.min(stat,axis=1))   #to get min number along the columns,ie,sideways
print(np.sum(stat))          #to get sum of matrix
print(np.sum(stat,axis=0))   #to get sum of number along the rows,ie,downward
print(np.sum(stat,axis=1))   #to get sum of number along the column,ie,sideways
```

```python
#Reorganising array

before=np.array([[1,2,3,4],[5,6,7,8]])
print(before)
print(before.reshape((8,1)))
print(before.reshape((4,2)))
print(before.reshape((2,2,2)))
print(before.reshape((2,3)))     #gives error as the nof elements in matrix != matrix elements of (2,3)
```

```python
#Vertical stacking         # no.of columns should be same

v1=np.array([1,2,3,4])
v2=np.array([5,6,7,8])
print(np.vstack([v1,v2]))
print(np.vstack([v1,v2,v1,v2]))
print(np.vstack((v2,v1)))       ##both " () " or " [] " can be used

# if v1= 1 2 3 4 5 , error would be shown
```

```python
#Horizontal stacking       # no.of rows shouls be same

h1=np.ones((2,4))
h2=np.zeros((2,6))
print(np.hstack((h1,h2)))
print(np.hstack((h2,h1)))
print(np.hstack([h1,h2,h1]))
```

```python
# Accessing from files

# This creates 'data.txt' inside Google Colab's cloud environment

with open('data.txt', 'w') as f:
    f.write("1,2,3\n4,5,6\n7,8,9")

print("File created on Google's server!")

# Now NumPy can read it perfectly

data=np.genfromtxt('data.txt')
print(data)
data = np.genfromtxt('data.txt', delimiter=',')
print(data)
print(data.astype('int'))      # to change the type of file content while printing or at moment
print(data.astype('int16'))
print(data.astype('int32'))
data=data.astype('int')        #to change the file type as int itself
print(data)
```

```python
# boolean masking and advanced indexing

print(data>2)
print(data[data>2])
print(data>=3)
print(np.any(data/2,axis=0))
print(np.any(np.any(data/2,axis=1)))
print(np.all(data>4))
print(np.all(data>4,axis=0))
print(np.all(data>4,axis=1))
```
